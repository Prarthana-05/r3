```r
# Create a data frame
painters <- data.frame(
  Name = c("Da Vinci", "Michelangelo", "Raphael",
           "Caravaggio", "Rembrandt"),
  Composition = c(15, 8, 17, 6, 18),
  Drawing = c(16, 17, 18, 6, 13),
  Colour = c(4, 4, 12, 16, 17),
  Expression = c(14, 8, 18, 0, 12),
  School = c("A", "A", "A", "E", "G")
)

# Display data frame
painters

# Summary
summary(painters)

# Summary of categorical variable
summary(painters$School)

# Summary of numeric variable
summary(painters$Composition)
```

```r
attach(painters)

# Access variables directly
summary(School)
summary(Composition)

# Detach data frame
detach(painters)
```

```r
# Painters belonging to School A
subset(painters, School == "A")

# Painters with Composition <= 10
subset(painters, Composition <= 10)

# Select only required columns
subset(painters, School == "A",
       select = c(Name, Composition, Drawing))
```

```r
# Split data according to School
splitted <- split(painters, painters$School)

# Display split data
splitted

# Check one group
splitted$A
```

```r
marks <- c(68, 82, 63, 86, 34,
           96, 41, 89, 29, 51,
           75, 77, 56, 59, 42)

# Calculate quantiles
quantile(marks)

# Specific probabilities
quantile(marks,
         probs = c(0, 0.20, 0.40,
                   0.60, 0.80, 1))
```
