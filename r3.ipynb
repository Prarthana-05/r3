### Running R code in a Python notebook using `rpy2`

First, we need to install the `rpy2` package. Then, we can load the `rpy2.ipython` extension to enable running R code in cells prefixed with `%%R`.

```python
pip install rpy2
```

```python
%load_ext rpy2.ipython
```

Now you can run R code in cells by starting them with `%%R`. Here's an example using some of your R code:

```python
%%R

# 1. Data Vectors
x <- c(2, 3, 5, 7)
y <- c(-2, -3, -5, 8)

cat("x =", x, "\n")
cat("y =", y, "\n\n")

# 2. Arithmetic Operations
cat("Addition:", x + y, "\n")
cat("Subtraction:", x - y, "\n")
cat("Multiplication:", x * y, "\n")
cat("Division:", x / y, "\n\n")

# 3. Power Operations
cat("Power:", x^2, "\n")
cat("Power with vector:", x^c(2, 3), "\n\n")

# 4. Integer Division
cat("Integer Division:", x %/% 2, "\n")
cat("Integer Division with vector:", x %/% c(2, 3), "\n\n")

# 5. Modulo Division
cat("Modulo:", x %% 2, "\n")
cat("Modulo with vector:", x %% c(2, 3), "\n\n")

# 6. Built-in Functions
cat("Maximum:", max(x), "\n")
cat("Minimum:", min(x), "\n")
cat("Mean:", mean(x), "\n")
cat("Absolute:", abs(c(-1, -2, -3, 4, 5)), "\n")
cat("Square Root:", sqrt(c(4, 9, 16, 25)), "\n")
cat("Sum:", sum(x), "\n")
cat("Product:", prod(x), "\n")
cat("Round:", round(1.83), "\n")
cat("Log:", log(10), "\n")
cat("Log10:", log10(100), "\n")
cat("Exponential:", exp(1), "\n\n")

# 7. Assignment
x1 <- c(1, 2, 3, 4)
x2 <- x1^2

cat("x1:", x1, "\n")
cat("x2:", x2, "\n\n")

# 8. Matrix Operations
m <- matrix(
  data = c(1, 2, 3, 4, 5, 6, 7, 8),
  nrow = 4,
  ncol = 2
)

cat("Matrix:\n")
print(m)

cat("\nElement [3,2]:", m[3,2], "\n")
cat("Dimensions:", dim(m), "\n")
cat("Rows:", nrow(m), "\n")
cat("Columns:", ncol(m), "\n")
cat("Mode:", mode(m), "\n")
cat("Attributes:\n")
print(attributes(m))

# 9. Row-wise Matrix
m2 <- matrix(
  data = c(1, 2, 3, 4, 5, 6, 7, 8),
  nrow = 4,
  ncol = 2,
  byrow = TRUE
)

cat("\nRow-wise Matrix:\n")
print(m2)
```
