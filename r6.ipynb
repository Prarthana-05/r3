Break command

```r
drink = c("coffee", "lemonade", "tea", "juice")
for (x in drink) {
if (x == "tea") {
break
}
print(x)
}
```

next command


```r
drink = c("coffee", "lemonade", "tea", "juice")
for (x in drink) {
if (x == "lemonade") {
next
}
print(x)
}
```

WHile loop ex-2

```r
sumfunction = function() {

  sum = 0

  number = as.integer(
    readline(prompt = "Please select any number less than 25: ")
  )

  while (number <= 25) {
    sum = sum + number
    number = number + 1
  }

  print(paste("The sum of numbers received from the While Loop:", sum))
}

sumfunction()
```
