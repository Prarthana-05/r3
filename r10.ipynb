# Part A – Strings: Display, Formatting, Substitution and Replacement

The notes introduce `sub()` for replacing the **first** occurrence of a substring and `gsub()` for replacing **all** occurrences. fileciteturn1file0L13-L21

## 2. `sub()` equivalent – replace the first occurrence

```python
# R:
# y = "Number of participants: 25"
# sub("25", "30", y)

y = "Number of participants: 25"

# Python equivalent of R sub(): replace only the first match
result = re.sub("25", "30", y, count=1)

print(result)
```

```python
# Example from the notes
y = "Mr. Singh is the smart one. Mr. Singh is funny, too."

# Replace only the first occurrence
result_sub = re.sub("Mr\\. Singh", "Professor Jha", y, count=1)

print(result_sub)
```

## 3. `gsub()` equivalent – replace all occurrences

```python
y = "Mr. Singh is the smart one. Mr. Singh is funny, too."

# Python equivalent of R gsub(): replace every matching occurrence
result_gsub = re.sub("Mr\\. Singh", "Professor Jha", y)

print(result_gsub)
```

## 4. `sub()` vs `gsub()` comparison

- `sub()` changes only the first matching occurrence.
- `gsub()` performs a global replacement.

This follows the distinction given in the notes. fileciteturn1file0L15-L20

```python
text = "Mr. Singh met Mr. Singh and then called Mr. Singh."

first_only = re.sub("Mr\\. Singh", "Professor Jha", text, count=1)
all_matches = re.sub("Mr\\. Singh", "Professor Jha", text)

print("Original:    ", text)
print("First only:  ", first_only)
print("All matches: ", all_matches)
```

## 5. Regular-expression searching – `grep()`

The notes explain that `grep()` searches a character vector for matches and returns the indices by default. With `value=TRUE`, it returns the matching elements instead. fileciteturn1file0L77-L86 fileciteturn1file0L98-L115

```python
strings = [
    "R Course",
    "exercises",
    "include examples of R language"
]

pattern = "ex"

# Python equivalent of grep(pattern, x):
# Return 0-based indices first, then convert to R-style 1-based indices.
matching_indices_python = [
    i for i, value in enumerate(strings)
    if re.search(pattern, value) is not None
]

matching_indices_r_style = [i + 1 for i in matching_indices_python]

print("Python indices (0-based):", matching_indices_python)
print("R-style indices (1-based):", matching_indices_r_style)

# Equivalent of grep(..., value=TRUE)
matching_values = [
    value for value in strings
    if re.search(pattern, value) is not None
]

print("Matching values:", matching_values)
```

### 5.1 Case-sensitive and case-insensitive search

```python
strings = [
    "R Course",
    "exercises",
    "include examples of r language",
    "in R software."
]

pattern = "R"

# R grep("R", str, ignore.case=FALSE, value=TRUE)
case_sensitive = [
    value for value in strings
    if re.search(pattern, value) is not None
]

# R grep("R", str, ignore.case=TRUE, value=TRUE)
case_insensitive = [
    value for value in strings
    if re.search(pattern, value, flags=re.IGNORECASE) is not None
]

print("Case-sensitive matches:")
print(case_sensitive)

print("\nCase-insensitive matches:")
print(case_insensitive)
```

```python
# Return indices, equivalent to value=FALSE
case_sensitive_indices = [
    i + 1 for i, value in enumerate(strings)
    if re.search("R", value) is not None
]

case_insensitive_indices = [
    i + 1 for i, value in enumerate(strings)
    if re.search("R", value, flags=re.IGNORECASE) is not None
]

print("Case-sensitive indices (R-style):", case_sensitive_indices)
print("Case-insensitive indices (R-style):", case_insensitive_indices)
```

### 5.2 Search the combined `x` and `y` example

```python
x = "R course 24.07.2021"
y = "Number of participants: 25"

combined = [x, y]

print("Combined strings:")
print(combined)

# grep("our", c(x, y))
our_indices = [
    i + 1 for i, value in enumerate(combined)
    if re.search("our", value) is not None
]

# grep("Num", c(x, y))
num_indices = [
    i + 1 for i, value in enumerate(combined)
    if re.search("Num", value) is not None
]

print("Indices containing 'our':", our_indices)
print("Indices containing 'Num':", num_indices)
```

## 6. `grepl()` equivalent

The notes describe `grepl()` as returning TRUE/FALSE values indicating whether a match is available. fileciteturn1file0L180-L186

```python
strings = [
    "R Course",
    "exercises",
    "include examples of R language"
]

# Equivalent of grepl("R", str)
grepl_R = [
    re.search("R", value) is not None
    for value in strings
]

print("grepl('R', strings):")
print(grepl_R)

# Equivalent of grepl("ex", str)
grepl_ex = [
    re.search("ex", value) is not None
    for value in strings
]

print("\ngrepl('ex', strings):")
print(grepl_ex)
```

# Part C – Combining and Merging Data Frames

The notes identify three main techniques:
- `cbind()` – combine columns side-by-side
- `merge()` – join using a common column
- `rbind()` – stack rows vertically fileciteturn1file0L680-L684

## 18. Create `df1` and `df2`

```python
df1 = pd.DataFrame({
    "state": ["UP", "MP", "AP", "JK"],
    "popnsize": [1000, 2000, 3000, 4000]
})

df2 = pd.DataFrame({
    "state": ["UP", "MP", "AP", "JK"],
    "samplesize": [100, 200, 300, 400],
    "surveycompleted": ["Yes", "No", "Yes", "No"]
})

print("df1:")
display(df1)

print("df2:")
display(df2)
```

## 19. `cbind()` equivalent

The notes show `cbind(df1, df2)` combining columns side-by-side. fileciteturn1file0L689-L697

```python
# pandas equivalent of cbind()
cbind_result = pd.concat(
    [df1.reset_index(drop=True), df2.reset_index(drop=True)],
    axis=1
)

cbind_result
```

## 20. `merge()` equivalent

The notes merge the two data frames using the common `state` column. fileciteturn1file0L731-L740

```python
# pandas equivalent of merge(df1, df2, by="state")
merge_result = pd.merge(df1, df2, on="state")

merge_result
```

```python
# Different merge types for practice
print("Inner merge:")
display(pd.merge(df1, df2, on="state", how="inner"))

print("Left merge:")
display(pd.merge(df1, df2, on="state", how="left"))

print("Right merge:")
display(pd.merge(df1, df2, on="state", how="right"))

print("Outer merge:")
display(pd.merge(df1, df2, on="state", how="outer"))
```

## 21. `rbind()` equivalent

The notes use `rbind()` to append one data frame below another. fileciteturn1file0L784-L790

```python
df11 = pd.DataFrame({
    "state": ["UP", "MP", "AP", "JK"],
    "popnsize": [1000, 2000, 3000, 4000]
})

df22 = pd.DataFrame({
    "state": ["Bihar", "Delhi", "Punjab"],
    "popnsize": [100, 200, 300]
})

print("df11:")
display(df11)

print("df22:")
display(df22)
```

```python
# pandas equivalent of rbind()
rbind_result = pd.concat(
    [df11, df22],
    ignore_index=True
)

rbind_result
```
