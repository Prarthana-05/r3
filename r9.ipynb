# Lecture 36 – Display and Formatting of Strings

The notes explain that R's `print()` is primarily for printing one object at a time, while `cat()` can concatenate multiple items and print them as continuous output. fileciteturn2file0L21-L36

Python's `print()` can directly print multiple objects, so it combines the roles illustrated by R's `print()` and `cat()`.

```python
# R:
# print("The zero occurs at", 2*pi, "radians.")

# Python equivalent
x = 2 * math.pi
print("The zero occurs at", x, "radians.")
```

## R `cat()` equivalent

R `cat()` places a space between arguments by default and uses `\n` for a newline. fileciteturn2file0L53-L58

In Python, `print()` provides similar behavior.

```python
print("The zero occurs at", 2 * math.pi, "radians.")
print()

# Explicit separator
print("The zero occurs at", 2 * math.pi, "radians.", sep=" ")
```

## Separator examples

The notes demonstrate `cat(x, sep=" ++ ")` and `cat(x, sep=" / ")`. fileciteturn2file0L103-L111

Python's `str.join()` is a close equivalent for joining vector elements with a separator.

```python
x = list(range(1, 11))

print(" ++ ".join(map(str, x)))
print(" / ".join(map(str, x)))
```

## Formatting numeric values

```python
x = 7

print("The square of", x, "is", x**2, "!")
print(f"The square of {x} is {x**2} !")

# R format(sqrt(x), digits=3) is approximated with Python numeric formatting.
print(f"The square root of {x} is approximately {math.sqrt(x):.3g}")
```

## Printing with labels

The notes show `cat()` with `fill=2` and labels generated from letters. fileciteturn2file0L144-L158

Python can reproduce the displayed idea with a loop.

```python
import string

x = list(range(1, 11))

for letter, value in zip(string.ascii_lowercase[:10], x):
    print(f"( {letter} ): {value}")
```

# Lecture 37 – `paste()` and `paste0()`

R `paste()` concatenates strings and returns a new string/vector, unlike `cat()`, whose main purpose is output. It supports `sep` and `collapse`. fileciteturn2file0L174-L200

Python equivalents:
- `str.join()` – joining strings with a separator
- f-strings / string concatenation – combining individual values
- list comprehensions – vectorized string construction


```python
# R: paste("Everybody", "loves", "R Programming.")
print(" ".join(["Everybody", "loves", "R Programming."]))

# R: paste(..., sep="*")
print("*".join(["Everybody", "loves", "R Programming."]))

# R: paste(..., sep="===")
print("===".join(["Everybody", "loves", "R Programming."]))
```

## Vectorized `paste()` example

```python
names = ["Prof. Singh", "Mr. Venkat", "Dr. Jha"]

result = [f"{name} is a good person." for name in names]
result
```

## `collapse` equivalent

R's `collapse` combines the generated vector into one string. fileciteturn2file0L257-L267

```python
collapsed = ", and ".join(result)
print(collapsed)
```

## `paste()` with numbers and custom separator

```python
parts = [1, " is first", 2, " is second", 3, " is third"]
print("#".join(map(str, parts)))

parts_with_spaces = [1, " is first ", 2, " is second ", 3, " is third "]
print("#".join(map(str, parts_with_spaces)))
```

## Creating `Ex_1`, `Ex_2`, ...

```python
x = [f"Ex_{i}" for i in range(1, 6)]
print(x)

print(x[0])  # R x[1]
print(x[1])  # R x[2]
print(x[2])  # R x[3]
print(x[4])  # R x[5]
```

### `collapse=""` equivalent

With R `paste("Ex", 1:5, sep="_", collapse="")`, the vector becomes one string. fileciteturn2file0L303-L323

```python
x_vector = [f"Ex_{i}" for i in range(1, 6)]
x_single = "".join(x_vector)

print("Vector:", x_vector)
print("Single string:", x_single)
```

## `paste0()` equivalent

The notes show that `paste0()` performs concatenation without the default space and supports vectorized combinations. fileciteturn2file0L328-L348

```python
# R paste0(1:10)
print([str(i) for i in range(1, 11)])

# R paste0(1:10, c("st","nd","rd",rep("th",7)))
suffixes = ["st", "nd", "rd"] + ["th"] * 7
ordinals = [f"{i}{suffix}" for i, suffix in zip(range(1, 11), suffixes)]
print(ordinals)

# R paste(1:10, ...)
ordinals_with_space = [f"{i} {suffix}" for i, suffix in zip(range(1, 11), suffixes)]
print(ordinals_with_space)
```

# Lecture 38 – String Splitting

R `strsplit(x, split, fixed=FALSE)` splits character-vector elements using a delimiter or regular expression. fileciteturn2file0L361-L370

Python equivalents:
- `str.split()` for ordinary separators
- `re.split()` when regular expressions are needed


```python
x = "The&!syntax&!of&!paste&!is!&available! &inthe online-help"

print("Split on '!':")
print(x.split("!"))

print("\nSplit on '&!':")
print(x.split("&!"))

print("\nSplit on '!&':")
print(x.split("!&"))
```

## Accessing individual split components

```python
y = x.split("!&")

print("Python first component:", y[0])
print("Python second component:", y[1])

# R y[[1]][3] would be NA for this example.
# Python raises IndexError if the element does not exist.
print("Number of components:", len(y))
```

## Splitting dates

The notes split dates such as `2020-07-24` into year, month and day. fileciteturn2file0L443-L456

```python
dates = ["2020-07-24", "2021-08-25", "2022-09-26", "2023-10-27"]

dates_split = [date.split("-") for date in dates]

for row in dates_split:
    print(row)
```

## Convert split dates to a character matrix

The R notes use `unlist()` and `matrix()` to create a 4×3 character matrix. fileciteturn2file0L460-L469

```python
date_matrix_text = np.array(dates_split, dtype=str)
print(date_matrix_text)
print("Shape:", date_matrix_text.shape)
```

## Convert split dates to a numeric matrix

The notes also convert the split values to numbers before creating the matrix. fileciteturn2file0L474-L484

```python
date_matrix_numeric = np.array(dates_split, dtype=int)
print(date_matrix_numeric)
print("Data type:", date_matrix_numeric.dtype)
```

# Lecture 39 – String Manipulations and Alphabets

The notes introduce `nchar()` for counting characters and `nzchar()` for testing whether strings are non-empty. fileciteturn2file0L512-L530

```python
x = "R course 24.07.2022"
y = "Number of participants: 25"

print("Length of x:", len(x))
print("Length of y:", len(y))
```

## `nzchar()` equivalent

```python
x = "R course 24.07.2022"
y = "Number of participants: 25"

print(bool(x))
print(bool(y))

values = ["Apple", "Banana", "Cake"]
print([bool(v) for v in values])

values2 = ["Apple", "", "Cake"]
print([bool(v) for v in values2])
```

## Character counts for vectors and numbers

The notes show `nchar()` applied to character vectors and numeric values after conversion to character strings. fileciteturn2file0L579-L591

```python
x = ["Apple", "Banana", "Cake"]
print([len(s) for s in x])

y = [2, 4, 6]
print([len(str(v)) for v in y])

z = [11, 222, 3333]
print([len(str(v)) for v in z])

z1 = [1.1, 2.22, 3.333]
print([len(str(v)) for v in z1])
```

## Lowercase and uppercase conversion

R's `tolower()` and `toupper()` convert alphabetic characters while leaving non-alphabetic characters unchanged. fileciteturn2file0L595-L610

Python equivalents are `.lower()` and `.upper()`.

```python
x = "R course will start from 24.07.2022"
print(x.upper())

z = "INDIAN INSTITUTE OF TECHNOLOGY"
print(z.lower())
```
