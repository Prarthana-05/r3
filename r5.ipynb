```r
# ==========================================================
# LECTURE 18: MISSING DATA HANDLING
# ==========================================================

cat("--- LECTURE 18: MISSING DATA HANDLING ---\n")

# 1. Detecting Missing Values
x <- NA
print(is.na(x))

x <- c(11, NA, 13, NA)
print(is.na(x))

# 2. Handling Missing Values in Calculations
print(mean(x))               # Returns NA
print(mean(x, na.rm = TRUE)) # Returns 12

# 3. Location and Count of NAs
print(which(is.na(x)))       # Indices: 2, 4
print(sum(is.na(x)))         # Total count: 2

# 4. Complete Cases and Omitting NAs
print(complete.cases(x))
y <- na.omit(x)
print(y)
print(mean(y))


# ==========================================================
# LECTURE 19: CONDITIONAL EXECUTIONS (if & if-else)
# ==========================================================

cat("\n--- LECTURE 19: CONDITIONAL EXECUTIONS ---\n")

# 1. Basic if statement
x <- 5
if (x > 4) print(x * 3)

x <- 3
if (x > 4) print(x * 3)

# Using {}
x <- 5
if (x > 4) { print(x * 3) }

# Using print() inside blocks
x <- 6
if (x > 3) {
  print("The value is more than 3")
}

x <- 2
if (x > 3) {
  print("The value is more than 3")
}

# 2. if-else statements
x <- 5
if (x == 3) { x <- x - 1 } else { x <- 2 * x }
print(x)

x <- 3
if (x == 3) { x <- x - 1 } else { x <- 2 * x }
print(x)

x <- 6
if (x > 3) {
  print("The value is more than 3")
} else {
  print("The value is less than 3")
}

x <- 2
if (x > 3) {
  print("The value is more than 3")
} else {
  print("The value is less than 3")
}


# ==========================================================
# LECTURE 20: NESTED if-else-if & ifelse()
# ==========================================================

cat("\n--- LECTURE 20: NESTED if-else-if & ifelse() ---\n")

# 1. Nested if-else-if function test
test_nested <- function(x) {
  if (x == 3) {
    x <- x - 1
  } else if (x < 3) {
    x <- x + 5
  } else {
    x <- 2 * x
  }
  return(x)
}

print(test_nested(5)) # Output: 10
print(test_nested(2)) # Output: 7
print(test_nested(3)) # Output: 2

# 2. Vectorized ifelse()
x <- 1:10
print(ifelse(x < 6, x^2, x + 1))

x <- c(7, 9, 8, 4)
print(ifelse(x %% 2 == 0, "even number", "odd number"))


# ==========================================================
# LECTURE 21: switch() AND which() FUNCTIONS
# ==========================================================

cat("\n--- LECTURE 21: switch() AND which() ---\n")

# 1. switch() function
print(switch(2, "apple", "banana", "orange"))
print(switch(1, "apple", "banana", "orange"))

print(switch("colour", "colour" = "blue", "gender" = "male", "volume" = 50))
print(switch("volume", "colour" = "blue", "gender" = "male", "volume" = 50))

# 2. which() on vectors
x <- c(10, 15, 8, 14, 6, 12)
print(which(x == 14))
print(which(x != 12))
print(which(x > 10))

# 3. which(), which.min(), and which.max() on matrices
mat <- matrix(nrow = 3, ncol = 3, data = 1:9)
print(mat)

print(which.min(mat))
print(which.max(mat))

print(which(mat %% 2 == 1))
print(which(mat %% 2 == 1, arr.ind = TRUE))
```
