```r
# ============================================================
# R PROGRAMMING - NPTEL WEEK 7
# Sequences, Dates, Alphabets, Repeats, Sorting,
# Ordering, Mode and Lists
# ============================================================


# ============================================================
# 1. SEQUENCES
# ============================================================

# Continuous sequences with increment/decrement of 1

1:10
10:1

5:15
15:5

-1:-10
-10:-1

-5:-15
-15:-5


# Decimal sequences using :

1.23:10
1.23:10.54
10.54:2.23

-1.23:-10
-5.23:6


# Using seq()

seq(10)

seq(1, 10)

seq(from = 1, to = 10)

seq(from = 1, to = 10, by = 2)

seq(from = 10, to = 1, by = -1)


# Sequence using a predefined variable

x <- 2
seq(1, x, x/10)

x <- 50
seq(0, x, x/10)


# Store sequence in a variable

x <- seq(1, 50, 1/2)
x

y <- 2 * x
y


# ============================================================
# 2. INDEX VECTOR
# ============================================================

x <- c(9, 8, 7, 6)

ind <- seq(along = x)

ind

# Accessing value using index vector
x[ind[2]]


# ============================================================
# 3. CURRENT DATE AND TIME
# ============================================================

Sys.time()

Sys.Date()


# ============================================================
# 4. SEQUENCES OF DATES
# ============================================================

# Sequence of first day of years

seq(
  as.Date("2010-01-01"),
  as.Date("2017-01-01"),
  by = "years"
)


# Sequence of days

seq(
  as.Date("2017-01-01"),
  by = "days",
  length.out = 6
)


# Sequence of months

seq(
  as.Date("2017-01-01"),
  by = "months",
  length.out = 6
)


# Sequence of years

seq(
  as.Date("2017-01-01"),
  by = "years",
  length.out = 6
)


# Sequence using start and end dates

startdate <- as.Date("2016-1-1")
enddate <- as.Date("2017-1-1")

out <- seq(
  enddate,
  startdate,
  by = "-1 month"
)

out


# ============================================================
# 5. LOWERCASE ALPHABETS
# ============================================================

letters

# Particular range

letters[1:3]

letters[3:1]

letters[21:23]

letters[2]


# ============================================================
# 6. UPPERCASE ALPHABETS
# ============================================================

LETTERS

# Particular range

LETTERS[1:3]

LETTERS[3:1]

LETTERS[21:23]

LETTERS[2]


# ============================================================
# 7. REPEATS - rep()
# ============================================================

# Repeat a value 10 times

rep(3.5, times = 10)


# Repeat vector as a whole

rep(1:4, 2)


# Create vector

x <- 1:4
x


# Repeat complete vector 3 times

rep(x, times = 3)


# Repeat every element 3 times

rep(x, each = 3)


# Repeat every element 2 times

rep(1:4, each = 2)


# Use both each and times

rep(1:4, each = 2, times = 3)

rep(1:4, times = 3, each = 2)


# ============================================================
# 8. DIFFERENT NUMBER OF REPETITIONS
# ============================================================

rep(1:4, 2:5)


# Using sequence for number of repetitions

ans <- seq(from = 2, to = 8, by = 2)

ans

rep(1:4, ans)


# ============================================================
# 9. REPEAT MATRIX
# ============================================================

x <- matrix(
  nrow = 2,
  ncol = 2,
  data = 1:4,
  byrow = TRUE
)

x

rep(x, 3)


# ============================================================
# 10. REPETITION OF CHARACTERS
# ============================================================

rep(c("a", "b", "c"), 2)

rep(c("apple", "banana", "cake"), 2)


# ============================================================
# 11. REPEAT FOR PRE-SPECIFIED LENGTH
# ============================================================

rep(2, length.out = 5)

rep(2, length = 5)

rep(c(2, 3), length = 5)

rep(c(2, 3, 4), length = 5)


# Character repetition with specified length

rep("apple", length = 5)

rep(c("a", "b", "c"), length = 2)

rep(c("a", "b", "c"), length = 5)


# ============================================================
# 12. SORTING
# ============================================================

y <- c(8, 5, 7, 6)

y

# Ascending order

sort(y)

# Descending order

sort(y, decreasing = TRUE)


# ============================================================
# 13. ORDERING
# ============================================================

y <- c(9, 8, 5, 7, 6)

y

# Positions of values in ascending order

order(y)

# Positions of values in descending order

order(y, decreasing = TRUE)


# You can also use order() to actually arrange the vector

y[order(y)]

y[order(y, decreasing = TRUE)]


# ============================================================
# 14. MODE
# ============================================================

# Numeric

mode(2.432)

mode(c(3, 4, 5, 6, 7, 8))


# Character

mode("India")

mode(c("India", "CANADA"))


# Factor

mode(factor(c("UP", "MP")))


# List

mode(list("India", "USA"))


# Data frame

mode(
  data.frame(
    x = 1:2,
    y = c("India", "USA")
  )
)


# Function

mode(print)


# ============================================================
# 15. LISTS
# ============================================================

# A list can contain different types of objects

mylist <- list(
  "India",
  25,
  c(1, 2, 3),
  TRUE
)

mylist


# Access list elements using [[ ]]

mylist[[1]]
mylist[[2]]
mylist[[3]]
mylist[[4]]


# ============================================================
# 16. MATRICES INSIDE LIST
# ============================================================

x1 <- matrix(
  nrow = 2,
  ncol = 2,
  data = 1:4,
  byrow = TRUE
)

x2 <- matrix(
  nrow = 2,
  ncol = 2,
  data = 5:8,
  byrow = TRUE
)

x1
x2

# Matrix addition

x1 + x2


# ============================================================
# 17. CHANGING MATRIX ELEMENT
# ============================================================

x1[2, 1] <- "hello"

x1

# This gives an error because x1 is no longer numeric

# x1 + x2


# ============================================================
# 18. LIST CONTAINING MATRICES
# ============================================================

# Recreate numeric matrices

x1 <- matrix(
  nrow = 2,
  ncol = 2,
  data = 1:4,
  byrow = TRUE
)

x2 <- matrix(
  nrow = 2,
  ncol = 2,
  data = 5:8,
  byrow = TRUE
)

# Create list

matlist <- list(x1, x2)

matlist


# Access first list element

matlist[1]

# Access second list element

matlist[2]


# ============================================================
# 19. LIST WITH DIFFERENT OBJECT TYPES
# ============================================================

z1 <- list(
  c("water", "juice", "lemonade"),
  rep(1:4, each = 2),
  matrix(
    data = 5:8,
    nrow = 2,
    ncol = 2,
    byrow = TRUE
  )
)

z1


# ============================================================
# 20. ACCESS ELEMENTS OF LIST USING [[ ]]
# ============================================================

z1[[1]]

z1[[2]]

z1[[3]]


# ============================================================
# 21. EXTRACT "juice" FROM THE FIRST LIST ELEMENT
# ============================================================

z1[[1]][2]


# Important difference:
# z1[1][2]       -> does NOT give "juice"
# z1[[1]][2]     -> gives "juice"

z1[1][2]

z1[[1]][2]


# ============================================================
# 22. NAMED LIST ELEMENTS
# ============================================================

student_list <- list(
  Name = "Prarthana",
  Age = 20,
  Marks = c(90, 95, 98)
)

student_list

# Access using name

student_list[["Name"]]

student_list$Name

student_list[["Marks"]]

student_list$Marks


# ============================================================
# 23. SUBLIST
# ============================================================

x <- list(
  "India",
  100,
  "R",
  c(1, 2, 3),
  TRUE
)

# Extract second and fifth elements

x[c(2, 5)]


# ============================================================
# END
# ============================================================
```
