Create Matrix

```r
x <- matrix(nrow=4, ncol=2, data=1:8, byrow=TRUE)

x
```

Rename rows and columns

```r
rownames(x) <- c("r1", "r2", "r3", "r4")
colnames(x) <- c("c1", "c2")

x
```

```r
x <- matrix(nrow=4, ncol=2, data=2)

x
```

Matrix addition

```r
x <- matrix(nrow=4, ncol=2, data=1:8, byrow=TRUE)

y <- matrix(nrow=4, ncol=2, data=11:18, byrow=TRUE)

x + y
```

Matrix multiplication

```r
x <- matrix(nrow=4, ncol=2, data=1:8, byrow=TRUE)

y <- matrix(nrow=2, ncol=4, data=11:18, byrow=TRUE)

x %*% y
```

Concatenate matrices row wise

```r
x <- matrix(nrow=3, ncol=2, data=1:6, byrow=TRUE)

y <- matrix(nrow=3, ncol=2, data=11:16, byrow=TRUE)

rbind(x, y)
```

Concatenate matrices column wise

```r
cbind(x, y)
```

Inverse matric

```r
y <- matrix(
  nrow=2,
  ncol=2,
  byrow=TRUE,
  data=c(84,100,100,120)
)

solve(y)
```

Eigen valuse

```r
y <- matrix(
  nrow=2,
  ncol=2,
  byrow=TRUE,
  data=c(84,100,100,120)
)

solve(y)
```
